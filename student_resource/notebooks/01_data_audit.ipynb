```python
def get_name_candidates(s1_name, name_index):
    return set(name_index.get(s1_name, []))


total_true_matches = 0
retrieved_true_matches = 0

for _, row in ground_truth.iterrows():

    s1_id = row["source1_entity_id"]
    true_ids = row["matched_entity_ids"].split(",")

    # No matches
    if true_ids == [""]:
        continue

    # Get S1 record
    s1_row = s1[s1["entity_id"] == s1_id].iloc[0]

    name_norm = s1_row["name_norm"]

    # Candidates from S2 + S3
    candidates = (
        get_name_candidates(name_norm, s2_name_index)
        |
        get_name_candidates(name_norm, s3_name_index)
    )

    true_ids_set = set(true_ids)

    total_true_matches += len(true_ids_set)
    retrieved_true_matches += len(
        true_ids_set.intersection(candidates)
    )

print("Total true matches:", total_true_matches)
print("Retrieved by name blocking:", retrieved_true_matches)

print(
    "Candidate Recall:",
    retrieved_true_matches / total_true_matches
)
```

```python
import unicodedata
import re

def normalize_name_v2(name):
    name = str(name).lower()

    # Remove accents / diacritics
    name = unicodedata.normalize("NFKD", name)
    name = "".join(
        char for char in name
        if not unicodedata.combining(char)
    )

    # Replace punctuation with spaces
    name = re.sub(r"[^a-z0-9]+", " ", name)

    # Normalize whitespace
    name = re.sub(r"\s+", " ", name).strip()

    return name
```

```python
s1["name_norm_v2"] = s1["business_name"].apply(normalize_name_v2)
s2["name_norm_v2"] = s2["business_name"].apply(normalize_name_v2)
s3["name_norm_v2"] = s3["business_name"].apply(normalize_name_v2)
```

```python
LEGAL_SUFFIXES = {
    "inc",
    "incorporated",
    "llc",
    "llp",
    "ltd",
    "limited",
    "corp",
    "corporation",
    "company",
    "co",
    "pvt",
    "private",
    "pvtltd",
    "plc",
    "gmbh",
    "ag",
    "bv",
    "nv",
    "pty"
}


def normalize_name_v3(name):
    # Start with V2 normalization
    clean = normalize_name_v2(name)

    if not clean:
        return ""

    # Remove legal entity terms
    tokens = clean.split()
    tokens = [token for token in tokens if token not in LEGAL_SUFFIXES]

    return " ".join(tokens)
```

```python
from unidecode import unidecode

def normalize_name_v4(name):
    # Transliterate Hindi, Tamil, Telugu, etc. to Latin characters
    name = unidecode(str(name))

    # Apply our V2 normalization
    name = normalize_name_v2(name)

    # Remove legal entity suffixes
    tokens = name.split()
    tokens = [
        token for token in tokens
        if token not in LEGAL_SUFFIXES
    ]

    return " ".join(tokens)
```

```python
def house_street_key(address):
    if not address:
        return ""

    text = str(address).lower()
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    tokens = text.split()

    if len(tokens) < 2:
        return ""

    # First token should be the house/building number
    if not tokens[0].isdigit():
        return ""

    house_number = tokens[0]
    street_token = tokens[1]

    return f"{house_number}:{street_token}"


s1["house_street_key"] = s1["business_address"].apply(house_street_key)
s2["house_street_key"] = s2["business_address"].apply(house_street_key)
s3["house_street_key"] = s3["business_address"].apply(house_street_key)

print(
    s1[["business_address", "house_street_key"]].head(10)
)
```

```python
# ============================================
# CITY + HOUSE NUMBER BLOCKING RECALL
# ============================================

# 1. Fast lookup dictionaries
# entity_id -> city_house_key

s1_city_house_lookup = (
    s1.set_index("entity_id")["city_house_key"]
    .to_dict()
)

s2_city_house_lookup = (
    s2.set_index("entity_id")["city_house_key"]
    .to_dict()
)

s3_city_house_lookup = (
    s3.set_index("entity_id")["city_house_key"]
    .to_dict()
)


# ============================================
# 2. Check every TRUE match
# ============================================

retrieved_count = 0
total_true_matches = 0

# Iterate using itertuples -> faster than iterrows()
for row in ground_truth.itertuples(index=False):

    s1_id = row.source1_entity_id

    # city+house key of S1
    s1_key = s1_city_house_lookup.get(s1_id, "")

    # Number of true matches
    total_true_matches += row.num_matches

    # If S1 has no city+house key,
    # this blocker cannot retrieve its matches
    if not s1_key:
        continue

    # All true S2/S3 matches
    matched_ids = row.matched_entity_ids.split(",")

    for matched_id in matched_ids:

        matched_id = matched_id.strip()

        # Determine whether this is S2 or S3
        if matched_id.startswith("S2-"):

            matched_key = s2_city_house_lookup.get(
                matched_id,
                ""
            )

        elif matched_id.startswith("S3-"):

            matched_key = s3_city_house_lookup.get(
                matched_id,
                ""
            )

        else:
            # Unexpected ID format
            continue

        # Blocking retrieves this true match
        # if both records have the same city+house key
        if matched_key and matched_key == s1_key:
            retrieved_count += 1


# ============================================
# 3. Calculate recall
# ============================================

recall = retrieved_count / total_true_matches


print("====================================")
print("CITY + HOUSE BLOCKING RESULTS")
print("====================================")

print("True matches retrieved:", retrieved_count)
print("Total true matches:", total_true_matches)

print(
    "Recall:",
    f"{recall * 100:.4f}%"
)
```

```python
# ============================================
# EXACT UNION CANDIDATE COUNT
# V3 NAME + HOUSE-STREET + CITY-HOUSE <= 1000
# ============================================

THRESHOLD = 1000


def prepare_counts(df, keys):
    return (
        df.loc[
            (df[keys] != "").all(axis=1),
            keys
        ]
        .value_counts()
        .reset_index(name="count")
    )


def product_count(s1_df, sx_df, keys):
    """
    Number of pairs sharing all keys.
    """

    a = prepare_counts(s1_df, keys)
    b = prepare_counts(sx_df, keys)

    merged = a.merge(
        b,
        on=keys,
        how="inner",
        suffixes=("_s1", "_sx")
    )

    return int(
        (merged["count_s1"] * merged["count_sx"]).sum()
    )


def single_block_count(s1_df, sx_df, key):
    """
    Number of candidate pairs for one blocking key.
    """

    a = (
        s1_df.loc[
            s1_df[key] != "",
            key
        ]
        .value_counts()
    )

    b = (
        sx_df.loc[
            sx_df[key] != "",
            key
        ]
        .value_counts()
    )

    common = a.index.intersection(b.index)

    return int(
        (a.loc[common] * b.loc[common]).sum()
    )


def city_house_count(s1_df, sx_df, threshold):
    """
    City+house candidates, ignoring empty keys
    and blocks larger than threshold.
    """

    a = (
        s1_df.loc[
            s1_df["city_house_key"] != "",
            "city_house_key"
        ]
        .value_counts()
    )

    b = (
        sx_df.loc[
            sx_df["city_house_key"] != "",
            "city_house_key"
        ]
        .value_counts()
    )

    common = a.index.intersection(b.index)

    # Keep only source-side blocks <= threshold
    valid = common[
        b.loc[common] <= threshold
    ]

    return int(
        (a.loc[valid] * b.loc[valid]).sum()
    )


def intersection_with_city_house(
    s1_df,
    sx_df,
    keys,
    threshold
):
    """
    Intersection with City+House.

    Example:
    keys = ["name_norm", "city_house_key"]

    Means a pair must share BOTH keys.
    """

    a = prepare_counts(s1_df, keys)
    b = prepare_counts(sx_df, keys)

    # Keep only city-house blocks <= threshold
    valid_keys = (
        b.groupby("city_house_key")["count"]
        .first()
    )

    valid_keys = valid_keys[
        valid_keys <= threshold
    ].index

    a = a[
        a["city_house_key"].isin(valid_keys)
    ]

    b = b[
        b["city_house_key"].isin(valid_keys)
    ]

    merged = a.merge(
        b,
        on=keys,
        how="inner",
        suffixes=("_s1", "_sx")
    )

    return int(
        (merged["count_s1"] * merged["count_sx"]).sum()
    )


def calculate_union(sx):

    # -------------------------
    # Individual blockers
    # -------------------------

    A = single_block_count(
        s1,
        sx,
        "name_norm"
    )

    B = single_block_count(
        s1,
        sx,
        "house_street_key"
    )

    C = city_house_count(
        s1,
        sx,
        THRESHOLD
    )


    # -------------------------
    # Pairwise intersections
    # -------------------------

    AB = product_count(
        s1,
        sx,
        ["name_norm", "house_street_key"]
    )

    AC = intersection_with_city_house(
        s1,
        sx,
        ["name_norm", "city_house_key"],
        THRESHOLD
    )

    BC = intersection_with_city_house(
        s1,
        sx,
        ["house_street_key", "city_house_key"],
        THRESHOLD
    )


    # -------------------------
    # Triple intersection
    # -------------------------

    ABC = intersection_with_city_house(
        s1,
        sx,
        [
            "name_norm",
            "house_street_key",
            "city_house_key"
        ],
        THRESHOLD
    )


    # Inclusion-exclusion
    union = (
        A + B + C
        - AB - AC - BC
        + ABC
    )

    return {
        "A_name": A,
        "B_house_street": B,
        "C_city_house": C,
        "AB": AB,
        "AC": AC,
        "BC": BC,
        "ABC": ABC,
        "union": union
    }


# ============================================
# S2
# ============================================

print("Calculating S2...")

s2_result = calculate_union(s2)

print("\nS2 RESULTS")
print(s2_result)


# ============================================
# S3
# ============================================

print("\nCalculating S3...")

s3_result = calculate_union(s3)

print("\nS3 RESULTS")
print(s3_result)


# ============================================
# FINAL
# ============================================

final_union = (
    s2_result["union"]
    + s3_result["union"]
)

print("\n====================================")
print("FINAL UNION CANDIDATE COUNT")
print("====================================")
print(
    f"City+House <= {THRESHOLD}"
)
print(
    f"S2 candidates: {s2_result['union']:,}"
)
print(
    f"S3 candidates: {s3_result['union']:,}"
)
print(
    f"TOTAL UNIQUE CANDIDATES: {final_union:,}"
)
```

```python
# ============================================
# RARE TOKEN - FAST TRUE MATCH RECALL
# ============================================

# S1 entity_id -> rare candidate IDs
s1_rare_candidate_index = {}

for s1_id, tokens in zip(
    s1["entity_id"],
    s1_tokens
):

    candidate_ids = set()

    for token in set(tokens):

        if token in s2_rare_index:
            candidate_ids.update(
                s2_rare_index[token]
            )

        if token in s3_rare_index:
            candidate_ids.update(
                s3_rare_index[token]
            )

    if candidate_ids:
        s1_rare_candidate_index[s1_id] = candidate_ids


print(
    "S1 with rare-token candidates:",
    len(s1_rare_candidate_index)
)


# ============================================
# TRUE MATCH RECALL
# ============================================

rare_token_retrieved = 0
total_true_matches = 0

for row in ground_truth.itertuples(index=False):

    s1_id = row.source1_entity_id

    total_true_matches += row.num_matches

    candidate_ids = s1_rare_candidate_index.get(
        s1_id,
        set()
    )

    if not candidate_ids:
        continue

    matched_ids = row.matched_entity_ids.split(",")

    for matched_id in matched_ids:

        matched_id = matched_id.strip()

        if matched_id in candidate_ids:
            rare_token_retrieved += 1


recall = (
    rare_token_retrieved
    / total_true_matches
)


print("\n====================================")
print("RARE TOKEN RECALL")
print("====================================")

print(
    "True matches retrieved:",
    rare_token_retrieved
)

print(
    "Total true matches:",
    total_true_matches
)

print(
    "Standalone recall:",
    f"{recall * 100:.4f}%"
)
```

```python
# ============================================
# RARE TOKEN THRESHOLD COMPARISON
# ============================================

thresholds = [10, 50]

for limit in thresholds:

    # Tokens allowed at this threshold
    s2_allowed = set(
        s2_address_token_counts[
            s2_address_token_counts <= limit
        ].index
    )

    s3_allowed = set(
        s3_address_token_counts[
            s3_address_token_counts <= limit
        ].index
    )

    total_candidates = 0
    candidate_counts = []

    for tokens in s1_tokens:

        candidate_ids = set()

        for token in set(tokens):

            if token in s2_allowed:
                candidate_ids.update(
                    s2_rare_index.get(token, set())
                )

            if token in s3_allowed:
                candidate_ids.update(
                    s3_rare_index.get(token, set())
                )

        count = len(candidate_ids)

        candidate_counts.append(count)
        total_candidates += count

    counts = pd.Series(candidate_counts)

    print("\n====================================")
    print(f"RARE TOKEN FREQUENCY <= {limit}")
    print("====================================")

    print(
        "Total candidates:",
        f"{total_candidates:,}"
    )

    print(
        "Average / S1:",
        f"{counts.mean():.2f}"
    )

    print(
        "Median / S1:",
        f"{counts.median():.0f}"
    )

    print(
        "95th percentile:",
        f"{counts.quantile(.95):.0f}"
    )

    print(
        "99th percentile:",
        f"{counts.quantile(.99):.0f}"
    )

    print(
        "Maximum:",
        f"{counts.max():.0f}"
    )
```

```python
# ============================================
# RARE TOKEN <=10 vs <=50
# INCREMENTAL RECALL
# ============================================

thresholds = [10, 50]

# Existing blockers
existing_retrieved = (
    v3_retrieved
    | house_street_retrieved
    | city_house_retrieved
)

TOTAL_TRUE = 7_638_365

print("Existing retrieved:", len(existing_retrieved))
print(
    "Existing recall:",
    f"{len(existing_retrieved) / TOTAL_TRUE * 100:.4f}%"
)


# --------------------------------------------
# Build allowed token sets
# --------------------------------------------

allowed_s2 = {}
allowed_s3 = {}

for limit in thresholds:

    allowed_s2[limit] = set(
        s2_address_token_counts[
            s2_address_token_counts <= limit
        ].index
    )

    allowed_s3[limit] = set(
        s3_address_token_counts[
            s3_address_token_counts <= limit
        ].index
    )


# --------------------------------------------
# Build candidate index for each threshold
# --------------------------------------------

threshold_candidate_indexes = {}

for limit in thresholds:

    candidate_index = {}

    s2_allowed = allowed_s2[limit]
    s3_allowed = allowed_s3[limit]

    for s1_id, tokens in zip(
        s1["entity_id"],
        s1_tokens
    ):

        candidate_ids = set()

        for token in set(tokens):

            if token in s2_allowed:
                candidate_ids.update(
                    s2_rare_index.get(token, set())
                )

            if token in s3_allowed:
                candidate_ids.update(
                    s3_rare_index.get(token, set())
                )

        if candidate_ids:
            candidate_index[s1_id] = candidate_ids

    threshold_candidate_indexes[limit] = candidate_index

    print(
        f"<= {limit}: S1 with candidates = "
        f"{len(candidate_index):,}"
    )


# --------------------------------------------
# Check true matches
# --------------------------------------------

new_matches = {
    10: set(),
    50: set()
}

for row in ground_truth.itertuples(index=False):

    s1_id = row.source1_entity_id

    matched_ids = row.matched_entity_ids.split(",")

    candidates_10 = threshold_candidate_indexes[10].get(
        s1_id, set()
    )

    candidates_50 = threshold_candidate_indexes[50].get(
        s1_id, set()
    )

    for matched_id in matched_ids:

        matched_id = matched_id.strip()

        pair = (s1_id, matched_id)

        # <=10
        if (
            matched_id in candidates_10
            and pair not in existing_retrieved
        ):
            new_matches[10].add(pair)

        # <=50
        if (
            matched_id in candidates_50
            and pair not in existing_retrieved
        ):
            new_matches[50].add(pair)


# --------------------------------------------
# Results
# --------------------------------------------

print("\n====================================")
print("RARE TOKEN INCREMENTAL RESULTS")
print("====================================")

for limit in thresholds:

    new_count = len(new_matches[limit])

    combined = len(existing_retrieved) + new_count

    recall = combined / TOTAL_TRUE * 100

    print(f"\nFrequency <= {limit}")

    print(
        "New matches:",
        f"{new_count:,}"
    )

    print(
        "Combined retrieved:",
        f"{combined:,}"
    )

    print(
        "Combined recall:",
        f"{recall:.4f}%"
    )
```

```python
# ============================================================
# FINAL BLOCKING RECALL
# Ground-truth matches vs generated candidates
# ============================================================

# Ground truth lookup
gt_lookup = ground_truth.set_index(
    "source1_entity_id"
)["matched_entity_ids"].to_dict()

total_true_matches = 0
retrieved_true_matches = 0
missed_matches = []

for i, row in candidates_df.iterrows():

    s1_id = row["source1_entity_id"]
    candidate_string = row["candidate_entity_ids"]

    # Ground-truth IDs
    gt_string = gt_lookup.get(s1_id, "")

    if not gt_string:
        continue

    true_ids = set(gt_string.split(","))

    total_true_matches += len(true_ids)

    # Candidate IDs
    if candidate_string:
        candidate_ids = set(candidate_string.split(","))
    else:
        candidate_ids = set()

    # Actual matches recovered by blocking
    recovered = true_ids & candidate_ids

    retrieved_true_matches += len(recovered)

    # Keep missed matches for analysis
    if len(recovered) < len(true_ids):
        missed_matches.append(
            (
                s1_id,
                true_ids - candidate_ids
            )
        )

print("====================================")
print("FINAL BLOCKING RECALL")
print("====================================")

print("Total true matches:", total_true_matches)
print("Retrieved true matches:", retrieved_true_matches)
print("Missed true matches:", total_true_matches - retrieved_true_matches)

recall = (
    retrieved_true_matches / total_true_matches
    if total_true_matches > 0 else 0
)

print(f"Blocking Recall: {recall:.4%}")

print("S1 records with missed matches:", len(missed_matches))
```

```python
# ============================================================
# AUDIT MISSED MATCHES
# Check whether missed true pairs SHOULD have been
# captured by any of our 4 blockers
# ============================================================

import random

# Fast entity lookups
s1_lookup = s1.set_index("entity_id").to_dict("index")
s2_lookup = s2.set_index("entity_id").to_dict("index")
s3_lookup = s3.set_index("entity_id").to_dict("index")

# Take a sample of missed pairs
sample_size = min(1000, len(missed_matches))

sample_missed = random.sample(
    missed_matches,
    sample_size
)

audit_results = []

for s1_id, missed_ids in sample_missed:

    s1_row = s1_lookup[s1_id]

    for target_id in missed_ids:

        if target_id.startswith("S2-"):
            target_row = s2_lookup.get(target_id)
        else:
            target_row = s3_lookup.get(target_id)

        if target_row is None:
            continue

        # ---------------------------------------------
        # BLOCKER 1: V3 NAME
        # ---------------------------------------------
        name_match = (
            s1_row["name_norm_v3"] != ""
            and
            s1_row["name_norm_v3"] == target_row["name_norm_v3"]
        )

        # ---------------------------------------------
        # BLOCKER 2: HOUSE + STREET
        # ---------------------------------------------
        hs_match = (
            s1_row["house_street_key"] != ""
            and
            s1_row["house_street_key"]
            == target_row["house_street_key"]
        )

        # ---------------------------------------------
        # BLOCKER 3: CITY + HOUSE
        # ---------------------------------------------
        ch_match = (
            s1_row["city_house_key"] != ""
            and
            s1_row["city_house_key"]
            == target_row["city_house_key"]
        )

        # ---------------------------------------------
        # BLOCKER 4: RARE ADDRESS TOKEN
        # ---------------------------------------------
        s1_tokens = set(s1_row["address_tokens"])

        target_tokens = set(target_row["address_tokens"])

        common_tokens = s1_tokens & target_tokens

        rare_token_match = False

        for token in common_tokens:

            if (
                token in s2_rare50_index
                or
                token in s3_rare50_index
            ):
                rare_token_match = True
                break

        audit_results.append({
            "s1": s1_id,
            "target": target_id,
            "name_match": name_match,
            "house_street_match": hs_match,
            "city_house_match": ch_match,
            "rare_token_match": rare_token_match
        })

audit_df = pd.DataFrame(audit_results)

print("Sampled missed pairs:", len(audit_df))

print("\n=== BLOCKER COVERAGE AMONG MISSED PAIRS ===")

print(
    "V3 name could capture:",
    audit_df["name_match"].sum()
)

print(
    "House+Street could capture:",
    audit_df["house_street_match"].sum()
)

print(
    "City+House could capture:",
    audit_df["city_house_match"].sum()
)

print(
    "Rare token could capture:",
    audit_df["rare_token_match"].sum()
)

# At least one blocker should theoretically capture
covered_by_any = (
    audit_df["name_match"]
    | audit_df["house_street_match"]
    | audit_df["city_house_match"]
    | audit_df["rare_token_match"]
)

print(
    "\nCould be captured by at least one blocker:",
    covered_by_any.sum()
)

print(
    "Genuine blocker gaps:",
    (~covered_by_any).sum()
)
```

```python
# ============================================================
# FAST MISSED-MATCH ANALYSIS
# ============================================================

import random
import pandas as pd

# Sample 500 missed S1 records
sample_n = min(500, len(missed_matches))

sample_missed = random.sample(
    missed_matches,
    sample_n
)

# Flatten missed pairs
rows = []

for s1_id, target_ids in sample_missed:
    for target_id in target_ids:
        rows.append((s1_id, target_id))

miss_df = pd.DataFrame(
    rows,
    columns=["s1_id", "target_id"]
)

print("Sampled S1 records:", sample_n)
print("Sampled missed pairs:", len(miss_df))


# ============================================================
# GET ONLY REQUIRED S1 RECORDS
# ============================================================

s1_small = s1[
    s1["entity_id"].isin(miss_df["s1_id"])
][[
    "entity_id",
    "name_norm_v3",
    "house_street_key",
    "city_house_key",
    "address_tokens"
]].rename(columns={
    "entity_id": "s1_id",
    "name_norm_v3": "s1_name",
    "house_street_key": "s1_hs",
    "city_house_key": "s1_ch",
    "address_tokens": "s1_tokens"
})


# ============================================================
# GET ONLY REQUIRED S2/S3 TARGETS
# ============================================================

s2_ids_needed = set(
    miss_df.loc[
        miss_df["target_id"].str.startswith("S2-"),
        "target_id"
    ]
)

s3_ids_needed = set(
    miss_df.loc[
        miss_df["target_id"].str.startswith("S3-"),
        "target_id"
    ]
)

s2_small = s2[
    s2["entity_id"].isin(s2_ids_needed)
][[
    "entity_id",
    "name_norm_v3",
    "house_street_key",
    "city_house_key",
    "address_tokens"
]].rename(columns={
    "entity_id": "target_id",
    "name_norm_v3": "target_name",
    "house_street_key": "target_hs",
    "city_house_key": "target_ch",
    "address_tokens": "target_tokens"
})

s3_small = s3[
    s3["entity_id"].isin(s3_ids_needed)
][[
    "entity_id",
    "name_norm_v3",
    "house_street_key",
    "city_house_key",
    "address_tokens"
]].rename(columns={
    "entity_id": "target_id",
    "name_norm_v3": "target_name",
    "house_street_key": "target_hs",
    "city_house_key": "target_ch",
    "address_tokens": "target_tokens"
})

targets_small = pd.concat(
    [s2_small, s3_small],
    ignore_index=True
)


# ============================================================
# MERGE ONLY THE SMALL SAMPLE
# ============================================================

audit = (
    miss_df
    .merge(s1_small, on="s1_id", how="left")
    .merge(targets_small, on="target_id", how="left")
)

print("Audit rows:", len(audit))


# ============================================================
# BLOCKER 1 — V3 NAME
# ============================================================

audit["name_match"] = (
    (audit["s1_name"] != "") &
    (audit["s1_name"] == audit["target_name"])
)


# ============================================================
# BLOCKER 2 — HOUSE + STREET
# ============================================================

audit["house_street_match"] = (
    (audit["s1_hs"] != "") &
    (audit["s1_hs"] == audit["target_hs"])
)


# ============================================================
# BLOCKER 3 — CITY + HOUSE
# ============================================================

audit["city_house_same"] = (
    (audit["s1_ch"] != "") &
    (audit["s1_ch"] == audit["target_ch"])
)


# ============================================================
# BLOCKER 4 — RARE ADDRESS TOKEN
# ============================================================

def check_rare_token(row):

    if not isinstance(row["s1_tokens"], list):
        return False

    if not isinstance(row["target_tokens"], list):
        return False

    common = (
        set(row["s1_tokens"])
        &
        set(row["target_tokens"])
    )

    if row["target_id"].startswith("S2-"):
        rare_index = s2_rare50_index
    else:
        rare_index = s3_rare50_index

    return any(
        token in rare_index
        for token in common
    )


audit["rare_token_match"] = audit.apply(
    check_rare_token,
    axis=1
)


# ============================================================
# FINAL ANALYSIS
# ============================================================

covered = (
    audit["name_match"]
    |
    audit["house_street_match"]
    |
    audit["city_house_same"]
    |
    audit["rare_token_match"]
)

print("\n====================================")
print("MISSED MATCH ANALYSIS")
print("====================================")

print("V3 Name:", audit["name_match"].sum())
print("House + Street:", audit["house_street_match"].sum())
print("City + House:", audit["city_house_same"].sum())
print("Rare Token:", audit["rare_token_match"].sum())

print("\nCovered by at least one existing blocker:",
      covered.sum())

print("Genuine blocker gaps:",
      (~covered).sum())
```
